# 03 — CUDA module (GPU runtime, T4)

**Runtime > Change runtime type > T4 GPU**.

Phase 10b. If `nvidia-smi` or `nvcc` is missing, stop with a clear message. Do not treat a confusing compile error as the result.

CMake currently rejects `-DUSE_CUDA=ON` until CUDA sources exist.

```python
%%bash
set -euo pipefail
if ! command -v nvidia-smi >/dev/null 2>&1; then
  echo "NO GPU: nvidia-smi not found. Change runtime to T4 GPU or skip this notebook."
  exit 0
fi
if ! nvidia-smi; then
  echo "NO GPU: nvidia-smi failed. Stop."
  exit 0
fi
if ! command -v nvcc >/dev/null 2>&1; then
  echo "NO CUDA: nvcc not found. Stop."
  exit 0
fi
nvcc --version
echo "GPU present, but CUDA kernels are not implemented until Phase 10b."
echo "STOP: do not configure -DUSE_CUDA=ON in Phase 1."
```
